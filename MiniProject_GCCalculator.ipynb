Mini Project: GC Calculator

```python
seq = 'ATGGCCATTGTAATGGGCCGCTGA'
```

```python
print(len(seq))
```

```python
print(seq.count('A'),seq.count('G'),seq.count('T'),seq.count('C'))
```

```python
gc = (seq.count('G') + seq.count('C'))/len(seq) * 100
```

```python
print(gc)
```

```python
print(f'GC Content: {gc:.1f}%')
```
