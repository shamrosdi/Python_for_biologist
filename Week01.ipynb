Arithmetic

```python
length = 1000
print(length / 3) # ordinary division
print(length // 3) # division without decimal place
print(length % 3) # bases left over. Modulator: it returns the remainder of the division
print(2**10) # power
```

String are sequences

```python
# string comes with a built-in tool 'len' that can help to count characters.
```

```python
seq = 'ATGCGTTAGCGGATCCTA'
```

```python
print(len(seq)) # to count the length of the characters in the variable
```

```python
print(seq.count('G')) # count the specific character in the sequence
```

```python
print(seq.lower()) # change the variable into lower case letter
```

```python
print(seq.replace('T','U')) # replace specific character in the sequence with another character
```

```python
print(seq.find('GGATCC')) # find specific position in the sequence
```

```python
rna = seq.replace('T', 'U') # methods never change the original string. They return a new one, so save the result in variable
```

Indexing and slicing

```python
# python counts position from 0, and a slice [start:end] stops just before end.
```

```python
print(seq[0]) # first base in the sequence
print(seq[-1]) # last base in the sequence
print(seq[0:3]) # positions 0, 1, 2
print(seq[:3])
print(seq[-3:]) # last three positions
print(seq[::-1]) # reversed
```

Formatted output

```python
# f-string lets you place values inside text. :.1f rounds to one decimal place
```

```python
gc = (seq.count('G') + seq.count('C'))/len(seq) * 100
```

```python
print(f'GC Content: {gc:.1f}%')
```
